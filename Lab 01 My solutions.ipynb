## My solution for problem 2

```python
plt.figure(figsize=(7,5))
plt.plot(x, y)

plt.xlim(-4, -2)
plt.ylim(0, 2)

plt.axhline(y=1, linestyle='--')

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Intersection Point 1 - Zoomed")
plt.grid(True)

plt.show()


plt.figure(figsize=(7,5))
plt.plot(x, y)

plt.xlim(4, 6)
plt.ylim(0, 2)

plt.axhline(y=1, linestyle='--')

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Intersection Point 2 - Zoomed")
plt.grid(True)

plt.show()

# Intersection points (to two decimal accuracy):
# x₁ ≈ -3.47
# x₂ ≈ 4.73
```
