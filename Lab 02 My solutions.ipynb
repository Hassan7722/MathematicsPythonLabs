```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 1000)

f = np.sqrt(3.3 * x)
g = (x - 6.02)**2

plt.figure(figsize=(8,6))

plt.plot(x, f, label=r"$f(x)=\sqrt{3.3x}$")
plt.plot(x, g, label=r"$g(x)=(x-6.02)^2$")

plt.title("Overview of the two functions")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()


plt.figure(figsize=(8,6))

plt.plot(x, f, label=r"$f(x)=\sqrt{3.3x}$")
plt.plot(x, g, label=r"$g(x)=(x-6.02)^2$")

plt.xlim(3.8,4.5)
plt.ylim(2.5,4.5)

plt.title("Zoomed view showing intersection")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()

plt.figure(figsize=(8,6))

plt.plot(x, f, label=r"$f(x)=\sqrt{3.3x}$")
plt.plot(x, g, label=r"$g(x)=(x-6.02)^2$")

plt.xlim(8,8.6)
plt.ylim(5.5,5)

plt.title("Zoomed view showing intersection")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()

```
