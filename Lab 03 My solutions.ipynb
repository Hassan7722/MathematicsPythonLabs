### My solution

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 2000)

y = (x >= -1.7) * (x < 1.3) * (1.7 + x) \
  + (x >= 1.3) * (x < 4.3) * (4.3 - x)

plt.figure(figsize=(8,5))

plt.plot(x, y, label='f(x)')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Piecewise Function')

plt.legend()
plt.grid(True)

plt.show()
```

### My solution

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-10, 10, 2000)

y = (x < -7.8) * (-837.05) \
  + (x >= -7.8) * (x < 7.3) * (x**3 - 8.13*x**2 - 4.14*x + 99.84) \
  + (x >= 7.3) * 25.39

plt.figure(figsize=(8,5))

plt.plot(x, y, label='f(x)')

plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Problem 2: Piecewise Function')

plt.legend()
plt.grid(True)

plt.show()
```
